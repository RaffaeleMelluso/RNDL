# Exercise: Diabetes Regression using PyTorch Neural Network

**Objective:**
In this exercise, you will build a neural network using PyTorch to predict diabetes progression based on a diabetes dataset. You will perform the following steps:

- Import the necessary libraries.
- Load the diabetes dataset.
- Set training parameters.
- Split the data into train and test sets and scale the data.
- Convert the data to PyTorch tensors and create DataLoader objects.
- Define the neural network model.
- Define a function to evaluate mean squared error (MSE).
- Create the model, loss function, and optimizer.
- Train the model and monitor training progress.
- Evaluate the model's performance on the test data.
- Plot the training and test MSE over epochs.

A well trained model should have a test MSE $\approx 3000$.
